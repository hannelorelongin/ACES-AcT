### Stage 3.1: conservation of Map-mediated lysine acetylation sites with FLAMS

In our research, we have characterized LUZ19 gp13/Map. On the basis of our SILAC experiment, it appears that Map acetylates several lysines on MetE and MetK. This can serve as a first indicator for potential functional relevance, and help us find relevant literature on these lysine acetylation sites. 

**Goals:**
* Analyze conservation of lysine acetylation sites in MetE and MetK.

* PROTEIN_POSITION.tsv: FLAMS output file, describing conserved PTMs detected at residue POSITION for the protein with UniProt identifier PROTEIN

#### Goal 1: Analyze conservation of lysine acetylation sites in MetE and MetK.

In this notebook, we will run FLAMS, a tool that searches PTM databases dbPTM, CPLM and SCOP3P, to facilitate the detection of conserved PTM sites. We will focus on the lysine acetylations detected on MetE and MetK, as described in our manuscript.

```python
# overview of the modifications on MetE and MetK, as described in the manuscript
modifications = [("P57703", 306), ("P57703", 738), ("Q9I5Z0", 285), ("Q9I5Z0", 386), ("Q9I5Z0", 35), ("Q9I5Z0", 220)]
```

```python
# run FLAMS for each of the modifications
for mod in modifications:
    uniprot_id, position = mod
    out = f"./a_FLAMS/{uniprot_id}_{position}.tsv"
    !FLAMS --id {uniprot_id} -p {position} --range 1 -o {out} -d ./a_FLAMS/data -e 0.001 -m acetylation
```

```python
# let's have a look at the output for these modifications
for mod in modifications:
    uniprot_id, position = mod
    out = f"./a_FLAMS/{uniprot_id}_{position}.tsv"
    data = pd.read_csv(out, sep="\t")
    num_matches = data.shape[0]
    species = data["Species"].str.split().str[:2].str.join(" ").tolist()
    num_species = len(set(species))
    print(f"Lysine acetylation in {uniprot_id} at K{position}: ")
    print(f"\tPreviously reported in {num_matches} proteins, in {num_species} species.")
    print(f"\tSpecies: {set(species)}")
```

After running FLAMS and taking a first look at the results, we can see that all lysine acetylation sites (detected on MetE and MetK in our manuscript) had been reported in other PTM studies. The majority of the PTMs have only been described in a limited number of species, however, two of the modifications stand out as frequently modified: MetE K738 and MetK K285.
